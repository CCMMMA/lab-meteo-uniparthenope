# Sensor quality control

Explore a sensor time series and detect missing values.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

if Path("sensor_data.csv").exists():
    data = pd.read_csv("sensor_data.csv", parse_dates=["time"])
else:
    data = pd.DataFrame({"time": pd.date_range("2026-11-13", periods=12, freq="h"),
                         "value": [10, 11, 12, None, 14, 13, 12, 11, None, 10, 9, 10]})

data.plot(x="time", y="value", title="Sensor time series")
plt.show()
print("Missing values:")
print(data.isnull().sum())
missing = data[data["value"].isnull()]
print("Detected gaps:", len(missing))
```
