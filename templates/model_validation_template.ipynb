# Model validation

Compare model output with observations loaded from CSV files.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

if Path("observations.csv").exists() and Path("model.csv").exists():
    observations = pd.read_csv("observations.csv")
    model = pd.read_csv("model.csv")
else:
    times = pd.date_range("2026-11-13", periods=24, freq="h")
    observations = pd.DataFrame({"time": times, "obs": range(24)})
    model = pd.DataFrame({"time": times, "model": [value * 1.05 for value in range(24)]})

data = pd.merge(observations, model, on="time")
plt.plot(data["time"], data["obs"], label="Observations")
plt.plot(data["time"], data["model"], label="Model")
plt.legend()
plt.title("Model–observation comparison")
plt.show()
```
