```python
# Sensor QC

Esplorazione delle serie temporali e rilevamento dei gap nei dati.


import pandas as pd
import matplotlib.pyplot as plt

# Carica dati (esempio)
df = pd.read_csv("sensor_data.csv", parse_dates=["time"])

# Visualizza dati
df.plot(x="time", y="value", title="Serie temporale")
plt.show()

# Controllo dati mancanti
print("Valori mancanti:")
print(df.isnull().sum())

# Gap detection semplice
missing = df[df["value"].isnull()]
print("Gap trovati:", len(missing))
```
