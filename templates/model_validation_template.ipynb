```python
# Model Validation

Confronto tra modello e osservazioni.

import pandas as pd
import matplotlib.pyplot as plt

# Dati esempio
obs = pd.read_csv("observations.csv")
model = pd.read_csv("model.csv")

# Merge
df = pd.merge(obs, model, on="time")

# Plot confronto
plt.plot(df["time"], df["obs"], label="Osservazioni")
plt.plot(df["time"], df["model"], label="Modello")
plt.legend()
plt.title("Confronto modello-osservazione")
plt.show()
```
